```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
input_path = Path("../data/processed/products_clean.csv")

if not input_path.exists():
    raise FileNotFoundError("Exécute d'abord 01_data_collection.ipynb.")

df = pd.read_csv(input_path, encoding="utf-8-sig")
print("Dimensions :", df.shape)
display(df.head())
```

```python
df_prep = df.copy()

print("Lignes :", len(df_prep))
print("Colonnes :", len(df_prep.columns))
```

```python
text_columns = df_prep.select_dtypes(include=["object", "string"]).columns

for column in text_columns:
    df_prep[column] = df_prep[column].astype("string").str.strip()

display(df_prep.head())
```

```python
numeric_columns = [
    "diameter_mm", "length_m", "pressure_rating_bar",
    "power_kw", "voltage_v", "price_dzd"
]

for column in numeric_columns:
    if column in df_prep.columns:
        df_prep[column] = pd.to_numeric(df_prep[column], errors="coerce")

df_prep[numeric_columns].info()
```

```python
missing = df_prep.isna().sum().sort_values(ascending=False)
display(missing)
```

```python
duplicates = df_prep[df_prep["product_id"].duplicated(keep=False)]

print("Lignes concernées par un identifiant dupliqué :", len(duplicates))
display(duplicates)
```

```python
for column in ["product_id", "name", "domain", "category"]:
    if column in df_prep.columns:
        print(column, "— valeurs manquantes :", df_prep[column].isna().sum())
```

```python
text_fields = [
    "name", "domain", "category", "subcategory",
    "brand", "model", "description", "material", "connection_type"
]
available_fields = [col for col in text_fields if col in df_prep.columns]

def combine_product_text(row):
    values = []
    for col in available_fields:
        value = row[col]
        if pd.notna(value) and str(value).strip():
            values.append(str(value).strip())
    return " ".join(values)

df_prep["content_text"] = df_prep.apply(combine_product_text, axis=1)

display(df_prep[["product_id", "content_text"]].head())
```

```python
if "last_verified" in df_prep.columns:
    df_prep["last_verified"] = pd.to_datetime(
        df_prep["last_verified"], errors="coerce"
    )
    display(df_prep[["product_id", "last_verified"]].head())
```

```python
output_path = Path("../data/processed/products_prepared.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)

df_prep.to_csv(output_path, index=False, encoding="utf-8-sig")

print("Fichier créé :", output_path.exists())
print("Chemin :", output_path)
```
